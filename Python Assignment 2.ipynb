# Experiment 2

# A. Reproducible Normalization Problem

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))

mean = X.mean()
std = X.std()

X_normalized = (X - mean) / std
normalized_mean = X_normalized.mean()
normalized_std = X_normalized.std()


print("X:", X_normalized)
print(" ")
print("Mean:", mean)
print(" ")
print("Std:", std)
print(" ")
print("Normalized Mean:", normalized_mean)
print(" ")
print("Normalized std:", normalized_std)
```

# B. Cubes Divisible by 4 Problem


```python
C = (np.arange(1,101)**3).reshape(10,10)
div_by_4 = C[C % 4 == 0]
print("Shape of C:", C.shape)
print(" ")
print("Array Divisible by 4:", div_by_4)
print(" ")
print("Number of selected elements:", len(div_by_4))
```
